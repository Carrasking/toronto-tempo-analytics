```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import matplotlib.gridspec as gridspec
import matplotlib.patheffects as pe
import numpy as np
```

```python
# ─────────────────────────────────────────
#  BRAND COLORS  (Toronto Tempo official palette)
# ─────────────────────────────────────────
NAVY   = "#0D1B2A"
RED    = "#C8102E"
GOLD   = "#E8C547"
CREAM  = "#F5F0E8"
STEEL  = "#8B9BB4"
WHITE  = "#FFFFFF"
DARK   = "#080F18"
MID    = "#1e3048"
```

```python
# ─────────────────────────────────────────
#  FIGURE & GRID
# ─────────────────────────────────────────
fig = plt.figure(figsize=(16, 11), facecolor=NAVY)
fig.patch.set_facecolor(NAVY)

gs = gridspec.GridSpec(
    3, 3,
    figure=fig,
    hspace=0.60, wspace=0.40,
    left=0.06, right=0.97,
    top=0.84, bottom=0.07
)
```

```python
# ─────────────────────────────────────────
#  HEADER
# ─────────────────────────────────────────
# Red top bar
fig.add_artist(mpatches.FancyBboxPatch(
    (0, 0.93), 1, 0.07,
    boxstyle="square,pad=0",
    facecolor=RED, edgecolor="none",
    transform=fig.transFigure, zorder=0
))

fig.text(0.5, 0.965, "TORONTO TEMPO  ·  2026 INAUGURAL SEASON  ·  ANALYTICS DASHBOARD",
         ha="center", va="center", fontsize=13, fontweight="bold",
         color=WHITE, zorder=1)

fig.text(0.06, 0.895, "Business Insights & Analytics  |  WNBA Expansion Team",
         ha="left", va="top", fontsize=9.5, color=STEEL)
fig.text(0.94, 0.895, "Jorge Carrasco  ·  Humber Polytechnic",
         ha="right", va="top", fontsize=9, color=STEEL, style="italic")

# Gold accent line under subtitle
fig.add_artist(plt.Line2D(
    [0.06, 0.94], [0.876, 0.876],
    transform=fig.transFigure,
    color=GOLD, linewidth=1.2
))
```

```python
# ─────────────────────────────────────────
#  PANEL 1 — Top Scorers (horizontal bars)
# ─────────────────────────────────────────
ax1 = fig.add_subplot(gs[0:2, 0])
ax1.set_facecolor(NAVY)
for spine in ax1.spines.values():
    spine.set_edgecolor(MID)

players = ["M. Mabrey", "B. Sykes", "K. Rice", "N. Sabally", "T. Fagbenle"]
ppg     = [19.8, 6.1, 2.6, 2.3, 2.0]
bar_colors = [RED, GOLD, STEEL, STEEL, STEEL]

y = np.arange(len(players))
bars = ax1.barh(y, ppg, color=bar_colors, height=0.52,
                zorder=3, edgecolor="none")

for bar, val in zip(bars, ppg):
    ax1.text(bar.get_width() + 0.4,
             bar.get_y() + bar.get_height() / 2,
             f"{val} ppg", va="center",
             fontsize=8.5, fontweight="bold", color=WHITE)

ax1.set_yticks(y)
ax1.set_yticklabels(players, fontsize=9.5, color=WHITE)
ax1.set_xlabel("Points Per Game", fontsize=8, color=STEEL, labelpad=6)
ax1.set_xlim(0, 27)
ax1.xaxis.grid(True, color=MID, linewidth=0.7, zorder=0)
ax1.set_axisbelow(True)
ax1.tick_params(axis="x", labelsize=7.5)

ax1.set_title("TOP SCORERS", fontsize=9, color=GOLD,
              fontweight="bold", pad=10, loc="left")

# Mabrey label
ax1.text(0.97, 0.94, "★ WNBA Record Holder",
         transform=ax1.transAxes, ha="right", va="top",
         fontsize=7, color=GOLD, style="italic")
```

```python
# ─────────────────────────────────────────
#  PANEL 2 — Win/Loss Donut
# ─────────────────────────────────────────
ax2 = fig.add_subplot(gs[0, 1])
ax2.set_facecolor(NAVY)
ax2.axis("equal")

wedge_props = dict(width=0.42, edgecolor=NAVY, linewidth=2.5)
wedges, _ = ax2.pie(
    [9, 9],
    colors=[RED, MID],
    startangle=90,
    wedgeprops=wedge_props
)

ax2.text(0, 0.14, "9–9", ha="center", va="center",
         fontsize=22, fontweight="bold", color=WHITE)
ax2.text(0, -0.22, "WINS  ·  LOSSES", ha="center", va="center",
         fontsize=7.5, color=STEEL)

ax2.set_title("SEASON RECORD", fontsize=9, color=GOLD,
              fontweight="bold", pad=10, loc="left")

ax2.text(0.5, -0.18,
         ".500 record as an expansion team —\nrare in any sport",
         transform=ax2.transAxes, ha="center", va="top",
         fontsize=7.5, color=STEEL, style="italic", linespacing=1.5)
```

```python
# ─────────────────────────────────────────
#  PANEL 3 — Record Milestones (text panel)
# ─────────────────────────────────────────
ax3 = fig.add_subplot(gs[0, 2])
ax3.set_facecolor(DARK)
ax3.axis("off")

# Border
ax3.add_patch(mpatches.FancyBboxPatch(
    (0.04, 0.04), 0.92, 0.90,
    boxstyle="round,pad=0.01",
    facecolor=DARK, edgecolor=RED, linewidth=1.8,
    transform=ax3.transAxes, zorder=0
))

ax3.text(0.5, 0.93, "RECORD-BREAKING MOMENTS",
         ha="center", fontsize=7.5, fontweight="bold",
         color=GOLD, transform=ax3.transAxes)

milestones = [
    ("53 pts",  "Mabrey — WNBA record tied"),
    ("125 pts", "Team high — all-time WNBA"),
    ("9×3s",    "Single-game record (×2)"),
    ("8,210",   "Opening Night sellout"),
    ("$50M",    "Expansion fee (Kilmer Sports)"),
]

for i, (val, desc) in enumerate(milestones):
    y_val = 0.80 - i * 0.168
    ax3.text(0.12, y_val, val,
             ha="left", fontsize=13, fontweight="bold",
             color=RED, transform=ax3.transAxes)
    ax3.text(0.12, y_val - 0.085, desc,
             ha="left", fontsize=7.8, color=STEEL,
             transform=ax3.transAxes)
    if i < len(milestones) - 1:
        ax3.plot([0.06, 0.94],
                 [y_val - 0.115, y_val - 0.115],
                 color=MID, linewidth=0.7,
                 transform=ax3.transAxes)

ax3.set_title("KEY NUMBERS", fontsize=9, color=GOLD,
              fontweight="bold", pad=10, loc="left")
```

```python
# ─────────────────────────────────────────
#  PANEL 4 — Mabrey Scoring Arc (line chart)
# ─────────────────────────────────────────
ax4 = fig.add_subplot(gs[1, 1:])
ax4.set_facecolor(NAVY)
for spine in ax4.spines.values():
    spine.set_edgecolor(MID)
ax4.tick_params(labelsize=7.5)

games  = list(range(1, 19))
scores = [18, 14, 22, 17, 28, 16, 20, 19,
          24, 21, 16, 25, 19, 22, 37, 18, 21, 53]
avg    = [19.8] * len(games)
wnba_avg = [10.5] * len(games)   # approx WNBA league avg for context

ax4.fill_between(games, scores, alpha=0.12, color=RED)
ax4.plot(games, scores, color=RED, linewidth=2.2, zorder=4,
         marker="o", markersize=4, markerfacecolor=RED,
         markeredgecolor=NAVY, markeredgewidth=0.8)
ax4.plot(games, avg, color=GOLD, linewidth=1.3,
         linestyle="--", label="Season avg (19.8 ppg)", zorder=3)
ax4.plot(games, wnba_avg, color=STEEL, linewidth=0.9,
         linestyle=":", label="WNBA avg (≈10.5 ppg)", zorder=2)

# Annotations
ax4.annotate("37 pts", xy=(15, 37), xytext=(12.2, 45),
             fontsize=8, color=GOLD, fontweight="bold",
             arrowprops=dict(arrowstyle="->", color=GOLD, lw=1.1))
ax4.annotate("53 pts\n★ WNBA Record", xy=(18, 53), xytext=(15.3, 58),
             fontsize=8, color=WHITE, fontweight="bold",
             arrowprops=dict(arrowstyle="->", color=WHITE, lw=1.1))

ax4.set_xlim(0.5, 18.5)
ax4.set_ylim(0, 68)
ax4.set_xlabel("Games played", fontsize=8, color=STEEL, labelpad=5)
ax4.set_ylabel("Points", fontsize=8, color=STEEL, labelpad=5)
ax4.yaxis.grid(True, color=MID, linewidth=0.6)
ax4.set_axisbelow(True)
ax4.legend(fontsize=7.5, facecolor=DARK, edgecolor=MID,
           labelcolor=WHITE, loc="upper left",
           framealpha=0.9)
ax4.set_title("MARINA MABREY — SCORING TRAJECTORY  (Games 1–18, 2026)",
              fontsize=9, color=GOLD, fontweight="bold",
              pad=10, loc="left")
```

```python
# ─────────────────────────────────────────
#  PANEL 5 — Business Insights (bottom row)
# ─────────────────────────────────────────
ax5 = fig.add_subplot(gs[2, :])
ax5.set_facecolor(DARK)
ax5.axis("off")

insights = [
    ("01  DRAFT VALUE",
     "Mabrey: expansion pick #6\n$2.4M / 2-year contract\n→ Top-10 WNBA scorer"),
    ("02  MARKET DEMAND",
     "8,500-seat venue: sold out\nfrom opening night\n→ Clear signal to scale"),
    ("03  BRAND EQUITY",
     "Owners: Serena Williams,\nMasai Ujiri, Geoff Molson\n→ Bilingual CA coverage"),
    ("04  COMPETITIVE KPI",
     ".500 record — Year 1\nCoach: Sandy Brondello\n→ Rare for expansion teams"),
]

n = len(insights)
box_w = 1 / n

for i, (title, body) in enumerate(insights):
    x0 = i / n
    highlight = (i == 0)

    rect = mpatches.FancyBboxPatch(
        (x0 + 0.012, 0.08), box_w - 0.028, 0.80,
        boxstyle="round,pad=0.01",
        facecolor=NAVY,
        edgecolor=RED if highlight else MID,
        linewidth=2.0 if highlight else 0.9,
        transform=ax5.transAxes, zorder=1
    )
    ax5.add_patch(rect)

    mid_x = x0 + box_w / 2
    ax5.text(mid_x, 0.80, title,
             ha="center", fontsize=8.2, fontweight="bold",
             color=GOLD, transform=ax5.transAxes)
    ax5.text(mid_x, 0.44, body,
             ha="center", fontsize=8.3, color=WHITE,
             transform=ax5.transAxes,
             linespacing=1.6, va="center")

ax5.set_title("BUSINESS INSIGHTS", fontsize=9, color=GOLD,
              fontweight="bold", pad=10, loc="left")
```
